```python
# 1. Use plt.bar() to create a vertical bar chart showing the number of daily orders for 5 popular food delivery apps (e.g., Zomato, Swiggy, Uber Eats, Dunzo, Domino's) and set a custom color for each bar.


import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]

orders = [150, 180, 100, 80, 120]

colors = ["red", "orange", "blue", "green", "purple"]

plt.bar(apps, orders, color=colors)

plt.title("Daily Orders by Food Delivery Apps")
plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

```python
# 2. Create a horizontal bar chart using plt.barh() to display the number of followers for 6 different Instagram influencers; add a title and label the x-axis as 'Followers (in millions)'.


import matplotlib.pyplot as plt

influencers = [
    "Influencer 1",
    "Influencer 2",
    "Influencer 3",
    "Influencer 4",
    "Influencer 5",
    "Influencer 6"
]

followers = [12, 8, 15, 10, 20, 6]

plt.barh(influencers, followers, color="skyblue")

plt.title("Instagram Influencers Followers")
plt.xlabel("Followers (in millions)")
plt.ylabel("Instagram Influencers")

plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.show()
```

```python
# 3. Modify your vertical bar chart to display the exact order count as a text label above each bar using plt.text().<br><br><em><strong>Hint:</strong> Use a for loop to iterate through the bars and place the text labels at the correct height.</em>


import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]

orders = [150, 180, 100, 80, 120]

colors = ["red", "orange", "blue", "green", "purple"]

bars = plt.bar(apps, orders, color=colors)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(height),
        ha="center",
        va="bottom"
    )

plt.title("Daily Orders by Food Delivery Apps")
plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")

plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
# 4. Change the width of the bars in your food delivery app chart to make them thinner, and adjust the spacing so the bars do not overlap.<br><br><em><strong>Constraint:</strong> Use the 'width' parameter in plt.bar() and experiment with values less than 0.8.</em>


import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]

orders = [150, 180, 100, 80, 120]

colors = ["red", "orange", "blue", "green", "purple"]

bars = plt.bar(
    apps,
    orders,
    color=colors,
    width=0.5
)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(height),
        ha="center",
        va="bottom"
    )

plt.title("Daily Orders by Food Delivery Apps")
plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")

plt.xticks(rotation=15)
plt.margins(x=0.1)

plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
