```python
# 1. Use plt.subplot() to create a figure with 2 rows and 1 column of plots showing your daily step count and water intake (in litres) for the last 7 days. Plot both as line charts, one in each subplot.

import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

steps = [4000, 5500, 6000, 7500, 8000, 9500, 10000]

water = [1.5, 2.0, 1.8, 2.5, 2.2, 3.0, 2.8]

plt.figure(figsize=(8, 7))

plt.subplot(2, 1, 1)
plt.plot(days, steps, color='blue', marker='o')
plt.title('Daily Step Count')
plt.xlabel('Days')
plt.ylabel('Number of Steps')
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(days, water, color='green', marker='o')
plt.title('Daily Water Intake')
plt.xlabel('Days')
plt.ylabel('Water Intake (Litres)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# 2. Create a 2x2 multi-panel figure using plt.subplots(). In each subplot, display the number of Instagram posts you made in each quarter (Q1, Q2, Q3, Q4) for the past year, using different chart types: line, bar, scatter, and pie.

import matplotlib.pyplot as plt

quarters = ['Q1', 'Q2', 'Q3', 'Q4']
posts = [15, 25, 20, 30]

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

ax[0, 0].plot(quarters, posts, color='blue', marker='o')
ax[0, 0].set_title('Quarterly Posts - Line Chart')
ax[0, 0].set_xlabel('Quarter')
ax[0, 0].set_ylabel('Number of Posts')
ax[0, 0].grid(True)

ax[0, 1].bar(quarters, posts, color='orange')
ax[0, 1].set_title('Quarterly Posts - Bar Chart')
ax[0, 1].set_xlabel('Quarter')
ax[0, 1].set_ylabel('Number of Posts')

ax[1, 0].scatter(quarters, posts, color='green', s=100)
ax[1, 0].set_title('Quarterly Posts - Scatter Chart')
ax[1, 0].set_xlabel('Quarter')
ax[1, 0].set_ylabel('Number of Posts')
ax[1, 0].grid(True)

ax[1, 1].pie(posts, labels=quarters, autopct='%1.1f%%')
ax[1, 1].set_title('Quarterly Posts - Pie Chart')

fig.suptitle('Instagram Quarterly Posts Dashboard')

plt.tight_layout()
plt.show()
```

```python
# 3. Make a grid of 1 row and 3 columns using plt.subplots(), and plot the number of food orders you placed on Zomato, Swiggy, and Domino's over the last 6 months, one app per subplot. Share the y-axis among all subplots so you can easily compare order counts across apps.<br><br><em><strong>Hint:</strong> Use the 'sharey' parameter in plt.subplots().</em>

import matplotlib.pyplot as plt

months = ['May', 'June', 'July', 'August', 'September', 'October']

zomato = [5, 7, 6, 9, 8, 10]
swiggy = [4, 6, 8, 7, 9, 11]
dominos = [2, 3, 4, 5, 4, 6]

fig, ax = plt.subplots(1, 3, figsize=(14, 5), sharey=True)

ax[0].plot(months, zomato, color='red', marker='o')
ax[0].set_title('Zomato Orders')
ax[0].set_xlabel('Months')
ax[0].set_ylabel('Number of Orders')
ax[0].grid(True)

ax[1].plot(months, swiggy, color='orange', marker='o')
ax[1].set_title('Swiggy Orders')
ax[1].set_xlabel('Months')
ax[1].grid(True)

ax[2].plot(months, dominos, color='blue', marker='o')
ax[2].set_title("Domino's Orders")
ax[2].set_xlabel('Months')
ax[2].grid(True)

fig.suptitle('Food Orders Comparison - Last 6 Months')

plt.tight_layout()
plt.show()
```
