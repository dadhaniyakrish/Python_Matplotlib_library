```python

# 1. Install Matplotlib in your Python environment and write a script named hello_plot.py that creates a simple line plot showing the number of daily steps you walked over the past 7 days.


import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

steps = [3000, 4500, 5000, 4000, 6500, 8000, 7000]

plt.plot(days, steps, marker="o")

plt.title("Daily Steps Over 7 Days")
plt.xlabel("Days")
plt.ylabel("Number of Steps")

plt.grid(True)
plt.show()
```

```python
# 3. Write a Python script that creates a Matplotlib figure with two Axes: one showing your daily Spotify listening hours for a week and the other showing your daily YouTube viewing hours for the same week. Display both plots in a single figure.


import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify_hours = [1, 2, 1.5, 2.5, 1, 3, 2]

youtube_hours = [2, 1.5, 2, 3, 2.5, 4, 3.5]

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].plot(days, spotify_hours, marker="o", color="green")
ax[0].set_title("Daily Spotify Listening Hours")
ax[0].set_xlabel("Days")
ax[0].set_ylabel("Hours")
ax[0].grid(True)

ax[1].plot(days, youtube_hours, marker="o", color="red")
ax[1].set_title("Daily YouTube Viewing Hours")
ax[1].set_xlabel("Days")
ax[1].set_ylabel("Hours")
ax[1].grid(True)

fig.suptitle("Weekly Spotify and YouTube Usage")
plt.tight_layout()
plt.show()
```

```python
# 4. Add axis labels, a title, and a grid to your plot from the previous task to clearly communicate the information, mimicking the style of analytics dashboards in apps like Instagram Insights.<br><br><em><strong>Hint:</strong> Use set_xlabel(), set_ylabel(), set_title(), and grid() methods on the Axes object.</em>


import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify_hours = [1, 2, 1.5, 2.5, 1, 3, 2]
youtube_hours = [2, 1.5, 2, 3, 2.5, 4, 3.5]

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].plot(days, spotify_hours, marker="o", color="green")
ax[0].set_title("Spotify Listening Insights")
ax[0].set_xlabel("Day of the Week")
ax[0].set_ylabel("Listening Hours")
ax[0].grid(True, linestyle="--", alpha=0.5)

ax[1].plot(days, youtube_hours, marker="o", color="red")
ax[1].set_title("YouTube Viewing Insights")
ax[1].set_xlabel("Day of the Week")
ax[1].set_ylabel("Viewing Hours")
ax[1].grid(True, linestyle="--", alpha=0.5)

fig.suptitle("Weekly Media Usage Dashboard")
plt.tight_layout()
plt.show()
```
