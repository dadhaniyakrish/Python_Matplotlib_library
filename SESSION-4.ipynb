```python
# 1. Use plt.pie() to visualize the percentage of time you spend on 5 different apps in a day (e.g., Instagram, YouTube, WhatsApp, Zomato, Spotify) and display the percentage values on each segment using the autopct parameter.


import matplotlib.pyplot as plt

apps = ["Instagram", "YouTube", "WhatsApp", "Zomato", "Spotify"]

time_spent = [3, 4, 2, 1, 2]

colors = ["purple", "red", "green", "orange", "blue"]

plt.pie(
    time_spent,
    labels=apps,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Daily App Usage Percentage")

plt.show()
```

```python
# 2. Create a pie chart showing the distribution of your monthly online spending across 4 categories: Food Delivery, Shopping, Entertainment, and UPI Payments, and use the explode parameter to highlight the category with the highest spending.<br><br><em><strong>Hint:</strong> Set a higher explode value for the segment you want to pull out from the pie.</em>


import matplotlib.pyplot as plt

categories = [
    "Food Delivery",
    "Shopping",
    "Entertainment",
    "UPI Payments"
]

spending = [2000, 5000, 1500, 3000]

explode = [0, 0.2, 0, 0]

colors = ["orange", "blue", "green", "red"]

plt.pie(
    spending,
    labels=categories,
    colors=colors,
    explode=explode,
    autopct="%1.1f%%",
    shadow=True,
    startangle=90
)

plt.title("Monthly Online Spending Distribution")

plt.show()
```

```python
# 3. Transform your pie chart into a donut chart to show the same spending data by adding a white circle at the center using plt.Circle and ax.add_artist.


import matplotlib.pyplot as plt

categories = [
    "Food Delivery",
    "Shopping",
    "Entertainment",
    "UPI Payments"
]

spending = [2000, 5000, 1500, 3000]

explode = [0, 0.2, 0, 0]

colors = ["orange", "blue", "green", "red"]

fig, ax = plt.subplots()

ax.pie(
    spending,
    labels=categories,
    colors=colors,
    explode=explode,
    autopct="%1.1f%%",
    startangle=90
)

centre_circle = plt.Circle(
    (0, 0),
    0.65,
    color="white"
)

ax.add_artist(centre_circle)

ax.set_title("Monthly Online Spending Donut Chart")

plt.show()
```

```python
# 4. Suppose you want to compare the market share of 4 IPL cricket teams based on the number of Instagram followers. Build a donut chart where each segment shows the team name and follower percentage, and the chart title clearly states 'IPL Teams Instagram Followers Share'.<br><br><em><strong>Constraint:</strong> Use both labels and autopct to display information on the chart.</em>


import matplotlib.pyplot as plt

teams = ["CSK", "MI", "RCB", "KKR"]

followers = [15, 14, 18, 10]

colors = ["yellow", "blue", "red", "purple"]

fig, ax = plt.subplots()

ax.pie(
    followers,
    labels=teams,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90
)

centre_circle = plt.Circle(
    (0, 0),
    0.65,
    color="white"
)

ax.add_artist(centre_circle)

ax.set_title("IPL Teams Instagram Followers Share")

plt.show()
```
