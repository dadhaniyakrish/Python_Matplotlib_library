```python
# 1. Use plt.plot() to create a line chart showing the number of daily steps you walked over the last 7 days. Set the line color to green and linewidth to 2.


import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

steps = [2000, 3500, 4000, 3000, 5000, 6500, 7000]

plt.plot(days, steps, color="green", linewidth=2)

plt.title("Daily Steps Over 7 Days")
plt.xlabel("Days")
plt.ylabel("Number of Steps")

plt.grid(True)

plt.show()
```

```python
# 2. Plot a line chart of your favorite music app's monthly active users for 6 months using plt.plot(). Add circle markers at each data point and set the marker color to red.<br><br><em><strong>Hint:</strong> Use the marker and markerfacecolor parameters in plt.plot().</em>


import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

users = [10, 15, 18, 20, 25, 30]

plt.plot(
    months,
    users,
    marker="o",
    markerfacecolor="red",
    color="blue",
    linewidth=2
)

plt.title("Spotify Monthly Active Users")
plt.xlabel("Months")
plt.ylabel("Active Users (Millions)")

plt.grid(True)

plt.show()
```

```python
# 3. Create a line chart comparing the number of orders you placed on Zomato and Swiggy over the last 5 months. Use different line styles and colors for each app, and add a legend to distinguish them.


import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

zomato_orders = [10, 15, 12, 20, 25]
swiggy_orders = [8, 12, 18, 15, 22]

plt.plot(
    months,
    zomato_orders,
    color="red",
    linestyle="-",
    marker="o",
    label="Zomato"
)

plt.plot(
    months,
    swiggy_orders,
    color="orange",
    linestyle="--",
    marker="s",
    label="Swiggy"
)

plt.title("Zomato vs Swiggy Monthly Orders")
plt.xlabel("Months")
plt.ylabel("Number of Orders")

plt.legend()
plt.grid(True)

plt.show()
```

```python
# 4. Add a main title 'Monthly Data Usage Trend' and label the X-axis as 'Month' and Y-axis as 'Data Used (GB)' to a line chart showing your mobile data usage for the past 6 months.<br><br><em><strong>Hint:</strong> Use plt.title(), plt.xlabel(), and plt.ylabel().</em>


import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

data_used = [15, 20, 18, 25, 30, 28]

plt.plot(months, data_used, color="blue", marker="o", linewidth=2)

plt.title("Monthly Data Usage Trend")

plt.xlabel("Month")
plt.ylabel("Data Used (GB)")

plt.grid(True)

plt.show()
```
