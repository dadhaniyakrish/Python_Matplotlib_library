```python
# 1. Create a line chart showing the number of daily active users on a music streaming app (like Spotify) over 7 days, and set the figure size to 10 inches wide by 4 inches tall using plt.figure().

import matplotlib.pyplot as plt

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Friday', 'Saturday', 'Sunday']

active_users = [1200, 1500, 1400, 1800, 2000, 2500, 2800]

plt.figure(figsize=(10, 4))

plt.plot(days, active_users, color='green', marker='o')

plt.title('Spotify Daily Active Users')
plt.xlabel('Days')
plt.ylabel('Number of Active Users')

plt.grid(True)
plt.show()
```

```python
# 2. Plot two lines on the same graph representing the number of food orders on Zomato and Swiggy over a week, using different line styles (e.g., dashed for Zomato, dotted for Swiggy) and add a legend to distinguish them.

import matplotlib.pyplot as plt

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Friday', 'Saturday', 'Sunday']

zomato_orders = [10, 15, 12, 18, 20, 25, 30]
swiggy_orders = [8, 12, 14, 16, 22, 28, 26]

plt.plot(days, zomato_orders,
         linestyle='--', color='red', marker='o',
         label='Zomato')

plt.plot(days, swiggy_orders,
         linestyle=':', color='orange', marker='s',
         label='Swiggy')

plt.title('Weekly Food Orders: Zomato vs Swiggy')
plt.xlabel('Days')
plt.ylabel('Number of Orders')

plt.legend()

plt.grid(True)
plt.show()
```

```python
# 3. Apply the 'ggplot' style to a chart showing Flipkart's daily sales for 10 days using plt.style.use('ggplot'), and customize the line color to orange.

import matplotlib.pyplot as plt

plt.style.use('ggplot')

days = range(1, 11)
sales = [12000, 15000, 13500, 18000, 20000,
         17000, 22000, 25000, 23000, 28000]

plt.plot(days, sales, color='orange', marker='o')

plt.title("Flipkart Daily Sales")
plt.xlabel("Days")
plt.ylabel("Sales Amount (₹)")

plt.show()
```

```python
# 4. Choose a color palette that matches Instagram's branding and plot a bar chart showing the number of posts uploaded each day for a week; use those colors for the bars.<br><br><em><strong>Hint:</strong> You can use a list of hex color codes to set bar colors.</em>

import matplotlib.pyplot as plt

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Friday', 'Saturday', 'Sunday']

posts = [2, 3, 1, 4, 2, 5, 3]

colors = [
    '#833AB4',  
    '#C13584', 
    '#E1306C',
    '#F77737',
    '#FCAF45',  
    '#C13584', 
    '#833AB4'   
]

plt.bar(days, posts, color=colors, edgecolor='black')

plt.title('Instagram Weekly Posts')
plt.xlabel('Days of the Week')
plt.ylabel('Number of Posts')

plt.show()
```

```python
# 5. Take any one of your previous charts and restyle it to match a dark theme using plt.style.use('dark_background'), then adjust the line or bar colors so that all text and data remain clearly visible.<br><br><em><strong>Constraint:</strong> Do not use the same colors as in the original chart.</em>

import matplotlib.pyplot as plt

plt.style.use('dark_background')

days = range(1, 11)

sales = [12000, 15000, 13500, 18000, 20000,
         17000, 22000, 25000, 23000, 28000]

plt.figure(figsize=(10, 5))

plt.plot(days, sales,
         color='cyan',
         marker='D',
         linewidth=2,
         markersize=7)

plt.title('Flipkart Daily Sales - Dark Theme')
plt.xlabel('Days')
plt.ylabel('Sales Amount (₹)')

plt.grid(True, alpha=0.3)
plt.show()
```
