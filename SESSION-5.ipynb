```python
# 1. Use plt.hist() to plot a histogram of delivery times (in minutes) for 30 Swiggy orders: [32, 28, 45, 30, 36, 40, 29, 34, 38, 31, 27, 41, 33, 35, 39, 37, 44, 30, 29, 32, 36, 28, 43, 31, 35, 40, 38, 33, 30, 42]. Show the plot window.


import matplotlib.pyplot as plt

delivery_times = [
    32, 28, 45, 30, 36, 40, 29, 34, 38, 31,
    27, 41, 33, 35, 39, 37, 44, 30, 29, 32,
    36, 28, 43, 31, 35, 40, 38, 33, 30, 42
]

plt.hist(delivery_times, bins=8, color="orange", edgecolor="black")

plt.title("Swiggy Delivery Time Distribution")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")

plt.grid(axis="y", alpha=0.5)

plt.show()
```

```python
# 2. Plot a histogram of IPL match scores (runs per team in a match) using plt.hist() and experiment with bins=5, bins=10, and bins=15. Observe and describe how changing the number of bins affects the shape of the histogram.<br><br><em><strong>Hint:</strong> Use any sample array of 50-60 scores between 120 and 220.</em>


import matplotlib.pyplot as plt

scores = [
    145, 160, 178, 190, 155, 200, 185, 170, 165, 210,
    150, 180, 195, 175, 160, 205, 188, 140, 220, 168,
    155, 190, 175, 200, 185, 165, 150, 210, 195, 180,
    170, 160, 205, 190, 155, 175, 185, 200, 145, 215,
    180, 165, 195, 170, 150, 205, 188, 160, 220, 175
]

fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].hist(scores, bins=5, color="blue", edgecolor="black")
ax[0].set_title("IPL Scores - 5 Bins")
ax[0].set_xlabel("Runs")
ax[0].set_ylabel("Frequency")

ax[1].hist(scores, bins=10, color="green", edgecolor="black")
ax[1].set_title("IPL Scores - 10 Bins")
ax[1].set_xlabel("Runs")
ax[1].set_ylabel("Frequency")

ax[2].hist(scores, bins=15, color="orange", edgecolor="black")
ax[2].set_title("IPL Scores - 15 Bins")
ax[2].set_xlabel("Runs")
ax[2].set_ylabel("Frequency")

fig.suptitle("IPL Match Scores Distribution")

plt.tight_layout()
plt.show()
```

```python
# 3. Given an array of daily step counts from a fitness tracker app for one month, plot two histograms using plt.hist(): one showing frequency (default), and one showing density=True. Compare the y-axis labels and explain the difference.


import matplotlib.pyplot as plt

steps = [
    3000, 4500, 5000, 6000, 7000,
    8000, 6500, 5500, 4000, 7500,
    9000, 8500, 6000, 7000, 5000,
    9500, 10000, 7500, 6500, 5500,
    4000, 8000, 9000, 7000, 6000,
    11000, 8500, 7500, 5000, 9500
]

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].hist(steps, bins=8, color="blue", edgecolor="black")
ax[0].set_title("Daily Steps - Frequency")
ax[0].set_xlabel("Number of Steps")
ax[0].set_ylabel("Frequency (Number of Days)")

ax[1].hist(
    steps,
    bins=8,
    density=True,
    color="orange",
    edgecolor="black"
)
ax[1].set_title("Daily Steps - Density")
ax[1].set_xlabel("Number of Steps")
ax[1].set_ylabel("Probability Density")

fig.suptitle("Monthly Daily Steps Distribution")

plt.tight_layout()
plt.show()
```

```python
# 4. Simulate the distribution of song durations (in seconds) from your Spotify playlist by generating 50 random numbers between 120 and 300. Plot a histogram using plt.hist() and set the color to 'purple' and add the title 'Spotify Song Duration Distribution'.

import numpy as np
import matplotlib.pyplot as plt

durations = np.random.randint(120, 301, 50)

plt.hist(durations, bins=10, color='purple', edgecolor='black')

plt.title('Spotify Song Duration Distribution')
plt.xlabel('Song Duration (Seconds)')
plt.ylabel('Number of Songs')

plt.show()
```
