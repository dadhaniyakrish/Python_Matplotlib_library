```python
# 1. Create a line chart using Matplotlib to show the number of daily active users for a music streaming app (like Spotify) over 12 months, and use plt.annotate() to label the month with the highest user count.

import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

active_users = [1200, 1500, 1800, 1600, 2200, 2500,
                2800, 3000, 2700, 3200, 3500, 4000]

max_users = max(active_users)
max_index = active_users.index(max_users)

plt.figure(figsize=(10, 5))
plt.plot(months, active_users, marker='o',
         color='green', linewidth=2)

plt.annotate(
    f'Highest: {max_users}',
    xy=(max_index, max_users),
    xytext=(max_index - 2, max_users - 500),
    arrowprops=dict(facecolor='red', arrowstyle='->'),
    fontsize=12,
    color='red'
)

plt.title('Spotify Monthly Active Users')
plt.xlabel('Months')
plt.ylabel('Daily Active Users')

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 2. Plot a bar chart showing the number of orders per month for a food delivery app (like Zomato), and add an arrow annotation using plt.annotate() to highlight the month with the lowest orders.<br><br><em><strong>Hint:</strong> Use the 'arrowprops' parameter to draw the arrow pointing to the bar.</em>

import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

orders = [1200, 1500, 900, 1800, 2200, 1700,
          2500, 2100, 1300, 2800, 3000, 2600]

min_orders = min(orders)
min_index = orders.index(min_orders)

plt.figure(figsize=(10, 5))
plt.bar(months, orders, color='orange')

plt.annotate(
    f'Lowest Orders: {min_orders}',
    xy=(min_index, min_orders),
    xytext=(min_index + 1, min_orders + 700),
    arrowprops=dict(facecolor='red', arrowstyle='->'),
    fontsize=12,
    color='red'
)

plt.title('Zomato Monthly Orders')
plt.xlabel('Months')
plt.ylabel('Number of Orders')

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# 3. Create a pie chart representing the percentage of different payment methods (UPI, Card, Wallet, Cash) used on an e-commerce app (like Flipkart), and add a legend positioned at the lower right corner of the plot.

import matplotlib.pyplot as plt

payment_methods = ['UPI', 'Card', 'Wallet', 'Cash']
percentages = [45, 25, 20, 10]

plt.figure(figsize=(8, 6))

plt.pie(
    percentages,
    labels=payment_methods,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Flipkart Payment Methods Distribution')

plt.legend(
    payment_methods,
    loc='lower right',
    title='Payment Methods'
)

plt.tight_layout()
plt.show()
```

```python
# 4. Given a scatter plot of movie ratings vs. number of reviews for movies on a streaming platform (like BookMyShow), use plt.annotate() to add a text box highlighting the movie with the highest rating.<br><br><em><strong>Hint:</strong> Set the 'bbox' parameter in plt.annotate() to create a visible text box.</em>

import matplotlib.pyplot as plt

movies = ['Movie A', 'Movie B', 'Movie C', 'Movie D',
          'Movie E', 'Movie F']

ratings = [7.5, 8.2, 9.5, 6.8, 8.7, 7.9]
reviews = [1200, 2500, 1800, 900, 3000, 1500]

max_rating = max(ratings)
max_index = ratings.index(max_rating)

plt.figure(figsize=(10, 6))

plt.scatter(reviews, ratings, color='blue', s=100)

plt.annotate(
    f'{movies[max_index]}\nRating: {max_rating}',
    xy=(reviews[max_index], max_rating),
    xytext=(reviews[max_index] + 300, max_rating - 0.5),
    arrowprops=dict(arrowstyle='->', color='red'),
    bbox=dict(boxstyle='round,pad=0.5',
              facecolor='yellow',
              edgecolor='black'),
    fontsize=11
)

plt.title('Movie Ratings vs Number of Reviews')
plt.xlabel('Number of Reviews')
plt.ylabel('Movie Ratings')

plt.grid(True)
plt.tight_layout()
plt.show()
```
