```python
# 1. Download a sample sales dataset (for example, from Kaggle or use a small CSV you create) with columns like product, category, date, and sales amount. Load it into a Pandas DataFrame and display the first 5 rows to verify the data.

import pandas as pd

data = {
    'Product': ['Laptop', 'Mobile', 'Keyboard', 'Mouse', 'Monitor',
                'Tablet', 'Headphones', 'Laptop', 'Mouse', 'Mobile'],

    'Category': ['Electronics', 'Electronics', 'Accessories',
                 'Accessories', 'Electronics', 'Electronics',
                 'Accessories', 'Electronics', 'Accessories',
                 'Electronics'],

    'Date': ['2026-01-01', '2026-01-02', '2026-01-03',
             '2026-01-04', '2026-01-05', '2026-01-06',
             '2026-01-07', '2026-01-08', '2026-01-09',
             '2026-01-10'],

    'Sales_Amount': [55000, 25000, 1500, 800, 12000,
                     18000, 2500, 55000, 800, 25000]
}

df = pd.DataFrame(data)

df.to_csv('sales_data.csv', index=False)
```

```python
# 2. Using Matplotlib, create a bar chart that shows total sales for each product category (like 'Electronics', 'Clothing', 'Groceries'), similar to how Flipkart or Myntra might track category performance.

import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Category': ['Electronics', 'Clothing', 'Groceries',
                 'Electronics', 'Clothing', 'Groceries'],
    'Sales': [50000, 30000, 20000, 40000, 25000, 15000]
}

df = pd.DataFrame(data)

category_sales = df.groupby('Category')['Sales'].sum()

plt.bar(category_sales.index, category_sales.values, color='skyblue')

plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales (₹)')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
# 3. Plot a line chart to visualize monthly sales trends over a year, using the date column to group sales by month. Add a title and axis labels to make the chart presentation-ready.

import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Date': pd.date_range(start='2026-01-01', periods=12, freq='MS'),
    'Sales': [25000, 30000, 28000, 35000, 40000, 38000,
              45000, 42000, 50000, 48000, 55000, 60000]
}

df = pd.DataFrame(data)

df['Date'] = pd.to_datetime(df['Date'])

monthly_sales = df.groupby(df['Date'].dt.month)['Sales'].sum()

plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index, monthly_sales.values,
         marker='o', color='blue', linewidth=2)

plt.title('Monthly Sales Trend - 2026')
plt.xlabel('Month')
plt.ylabel('Total Sales (₹)')

plt.xticks(range(1, 13),
           ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
            'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])

plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# 5. Add value annotations to your category bar chart to display the exact sales amount above each bar, then export the chart as a PNG image file named 'category_sales_kpi.png'.

import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Category': ['Electronics', 'Clothing', 'Groceries',
                 'Electronics', 'Clothing', 'Groceries'],
    'Sales': [50000, 30000, 20000, 40000, 25000, 15000]
}

df = pd.DataFrame(data)

category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(9, 5))
bars = plt.bar(category_sales.index, category_sales.values,
               color='skyblue', edgecolor='black')

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f'₹{height:,.0f}',
        ha='center',
        va='bottom'
    )

plt.title('Category-wise Sales KPI')
plt.xlabel('Product Category')
plt.ylabel('Total Sales (₹)')

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()

plt.savefig('category_sales_kpi.png', dpi=300, bbox_inches='tight')

plt.show()
```
