```python
# 1. Create a simple bar chart showing the number of orders for 5 different Zomato restaurants and save the chart as a PNG file using plt.savefig() with a DPI of 150.

import matplotlib.pyplot as plt

restaurants = ['Restaurant A', 'Restaurant B',
               'Restaurant C', 'Restaurant D',
               'Restaurant E']

orders = [1200, 1800, 900, 2200, 1500]

plt.figure(figsize=(9, 5))
plt.bar(restaurants, orders, color='orange')

plt.title('Zomato Restaurant Orders')
plt.xlabel('Restaurants')
plt.ylabel('Number of Orders')

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.savefig('zomato_orders.png', dpi=150)

plt.show()
```

```python
# 2. Generate a line plot that displays your daily step count for one week (7 days) and export it as an SVG file using plt.savefig().<br><br><em><strong>Hint:</strong> Change the file extension in the filename argument to save as SVG.</em>

import matplotlib.pyplot as plt

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Friday', 'Saturday', 'Sunday']

steps = [4500, 6000, 7500, 5000, 8500, 10000, 9000]

plt.figure(figsize=(10, 5))
plt.plot(days, steps, marker='o', color='green',
         linewidth=2, label='Daily Steps')

plt.title('Daily Step Count for One Week')
plt.xlabel('Days')
plt.ylabel('Number of Steps')

plt.grid(True)
plt.legend()

plt.savefig('daily_steps.svg', format='svg')

plt.show()
```

```python
# 3. Build a pie chart showing the percentage of time you spend on 4 different apps (Instagram, YouTube, WhatsApp, Spotify) in a day and save the chart as a PDF using plt.savefig().

import matplotlib.pyplot as plt

apps = ['Instagram', 'YouTube', 'WhatsApp', 'Spotify']

usage = [3, 4, 2, 1]

plt.figure(figsize=(8, 6))

plt.pie(
    usage,
    labels=apps,
    autopct='%1.1f%%',
    startangle=90,
    shadow=True
)

plt.title('Daily App Usage Distribution')

plt.legend(apps, title='Apps', loc='upper right')

plt.savefig('app_usage.pdf', format='pdf')

plt.show()
```
