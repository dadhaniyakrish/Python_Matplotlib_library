```python
# 2. Create a scatter plot showing 'amount spent on Zomato orders' vs 'number of orders placed' for 12 friends, and label each point with the friend's name.<br><br><em><strong>Hint:</strong> Use the 'annotate' function in matplotlib to add labels.</em>

import matplotlib.pyplot as plt

names = [
    "Amit", "Rahul", "Krish", "Jay",
    "Ravi", "Neha", "Priya", "Raj",
    "Karan", "Meera", "Dev", "Riya"
]

amount_spent = [1200, 2500, 1800, 3000, 900, 2200,
                1500, 3500, 2800, 1700, 4000, 1100]

orders = [4, 8, 6, 10, 3, 7, 5, 12, 9, 6, 14, 4]

plt.scatter(orders, amount_spent, color='purple', marker='o')

for i in range(len(names)):
    plt.annotate(
        names[i],
        (orders[i], amount_spent[i]),
        xytext=(5, 5),
        textcoords='offset points'
    )

plt.title('Zomato Spending vs Number of Orders')
plt.xlabel('Number of Orders Placed')
plt.ylabel('Amount Spent (₹)')

plt.grid(True)
plt.show()
```

```python
# 3. Add a trend line to your Zomato scatter plot by calculating the best-fit line using numpy's polyfit(), then plot it on top of the scatter plot.<br><br><em><strong>Hint:</strong> Use np.polyfit() to get the slope and intercept, then plot the line with plt.plot().</em>

import numpy as np
import matplotlib.pyplot as plt

names = [
    "Amit", "Rahul", "Krish", "Jay",
    "Ravi", "Neha", "Priya", "Raj",
    "Karan", "Meera", "Dev", "Riya"
]

orders = [4, 8, 6, 10, 3, 7, 5, 12, 9, 6, 14, 4]

amount_spent = [1200, 2500, 1800, 3000, 900, 2200,
                1500, 3500, 2800, 1700, 4000, 1100]

plt.scatter(orders, amount_spent, color='purple', label='Friends')

for i in range(len(names)):
    plt.annotate(names[i], (orders[i], amount_spent[i]),
                 xytext=(5, 5), textcoords='offset points')

slope, intercept = np.polyfit(orders, amount_spent, 1)

trend_line = np.poly1d([slope, intercept])

x = np.linspace(min(orders), max(orders), 100)
plt.plot(x, trend_line(x), color='red', label='Trend Line')

plt.title('Zomato Spending vs Number of Orders')
plt.xlabel('Number of Orders Placed')
plt.ylabel('Amount Spent (₹)')

plt.legend()
plt.grid(True)
plt.show()
```

```python
# 4. Given a dataset of 'YouTube watch hours' and 'Spotify listening hours' for 15 users, use numpy polyfit to fit a regression line, plot both the scatter and the trend line, and print the equation of the line on the plot.

import numpy as np
import matplotlib.pyplot as plt

youtube_hours = [2, 3, 4, 5, 2, 6, 7, 3, 8, 4, 5, 9, 6, 7, 10]

spotify_hours = [3, 4, 5, 6, 3, 7, 8, 4, 9, 5, 6, 10, 7, 8, 11]

slope, intercept = np.polyfit(youtube_hours, spotify_hours, 1)

regression_line = np.poly1d([slope, intercept])

x = np.linspace(min(youtube_hours), max(youtube_hours), 100)
y = regression_line(x)

plt.scatter(youtube_hours, spotify_hours,
            color='purple', label='Users')

plt.plot(x, y, color='red', label='Regression Line')

equation = f'y = {slope:.2f}x + {intercept:.2f}'
plt.text(2, 10, equation, fontsize=11, color='red')

plt.title('YouTube Watch Hours vs Spotify Listening Hours')
plt.xlabel('YouTube Watch Hours')
plt.ylabel('Spotify Listening Hours')

plt.legend()
plt.grid(True)
plt.show()

print("Regression Equation:", equation)
```
